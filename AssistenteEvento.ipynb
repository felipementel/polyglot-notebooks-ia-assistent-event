```polyglot-notebook
DotEnv.Load();
string model = Environment.GetEnvironmentVariable("model");
string deploymentName = Environment.GetEnvironmentVariable("deploymentName");
string apiKey = Environment.GetEnvironmentVariable("apiKey");
string URI = Environment.GetEnvironmentVariable("uri");
Console.WriteLine(model);
Console.WriteLine(deploymentName);
Console.WriteLine(apiKey);
Console.WriteLine(URI);
```

```polyglot-notebook
Azure.AI.OpenAI.AzureOpenAIClientOptions options = new()
{
    Audience = AzureOpenAIAudience.AzurePublicCloud,
};
AzureOpenAIClient azureClient = new(
    new Uri(URI),
    new Azure.AzureKeyCredential(apiKey),
    options);
```

```polyglot-notebook
OpenAI.Chat.ChatClient chatClient = azureClient.GetChatClient(deploymentName);
```

```polyglot-notebook
var requestOptions = new OpenAI.Chat.ChatCompletionOptions()
{
    TopP = 0.9f,
};
```

```polyglot-notebook
List<OpenAI.Chat.ChatMessage> messages = new List<OpenAI.Chat.ChatMessage>()
{
    new OpenAI.Chat.SystemChatMessage(systemPrompt)
};

var response = chatClient.CompleteChat(messages, requestOptions);
//Console.Write(response.Value.Content[0].Text);
```

```polyglot-notebook
var chatting = true;

//Console.WriteLine(systemPrompt);
int interation = 1;
while (chatting)
{
    var userMessage = await Kernel.GetInputAsync();

    if (userMessage.ToLower() == "sair")
    {
        chatting = false;
        break;
    }

    Console.WriteLine($" ======i-{interation}====== ");
    Console.WriteLine("Usuário: " + userMessage);

    messages.Add(new OpenAI.Chat.UserChatMessage(userMessage));
    var response = chatClient.CompleteChat(messages, requestOptions);
    System.Console.WriteLine();

    Console.WriteLine();
    Console.WriteLine("Assistente: " + response.Value.Content[0].Text);
    Console.WriteLine($" ======f-{interation}======  ");

    interation++;
}
```
